```python
# Function to detect outliers using IQR method
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (data[column] < lower_bound) | (data[column] > upper_bound)

print("🔍 Detecting outliers using IQR method...")

# Check for outliers in numeric columns
numeric_cols = ['age', 'salary', 'experience_years', 'performance_score', 'bonus_percentage']
outliers_dict = {}

for col in numeric_cols:
    outliers = detect_outliers_iqr(df_clean, col)
    outliers_dict[col] = outliers.sum()
    print(f"{col}: {outliers.sum()} outliers")

print("\n" + "="*80 + "\n")

# Fix invalid values based on domain knowledge
print("🔧 Fixing invalid values...")

# Age should be between 18 and 70
df_clean.loc[df_clean['age'] > 70, 'age'] = df_clean[df_clean['age'] <= 70]['age'].median()
df_clean.loc[df_clean['age'] < 18, 'age'] = df_clean[df_clean['age'] >= 18]['age'].min()

# Performance score should be 1-5
df_clean['performance_score'] = df_clean['performance_score'].clip(1, 5)

# Bonus percentage should be 0-20
df_clean['bonus_percentage'] = df_clean['bonus_percentage'].clip(0, 20)

print("✅ Invalid values fixed!")
print(f"\nAge range: {df_clean['age'].min()} - {df_clean['age'].max()}")
print(f"Performance score range: {df_clean['performance_score'].min()} - {df_clean['performance_score'].max()}")
print(f"Bonus percentage range: {df_clean['bonus_percentage'].min()} - {df_clean['bonus_percentage'].max()}")
```
